```r
allreads<-data.table()
C=0
for (i in list.files(path = "/hpcdata/lvd_qve/Projects/PTD_StrandSpecificCounting_scRNAseq/SCISSORS_2/",pattern = "CBC.csv",full.names = TRUE)){
    C=C+1
    print(i)
    label=strsplit(strsplit(i,"/hpcdata/lvd_qve/Projects/PTD_StrandSpecificCounting_scRNAseq/SCISSORS_2//")[[1]][2],split = "_S")[[1]][1]
    print(label)
    SAM=fread(i,nThread = 12)
    SAM[,sample:=i]
    SAM[,datalabel:=label]
    SAM[,UMIcount:=length(unique(UMI)),by="CBC"]
    SAM[,strand:=ifelse(flag==16,"Positive","Negative")]
    #SAM<-merge(SAM,BC,by.x = "CBC",by.y = "Barcode")
    counts=dcast.data.table(SAM,CBC+sample+datalabel+strand~ref_name,value.var = "UMI",fun.aggregate = function(U){length(unique(U))})
    allreads=rbindlist(fill = TRUE,list(allreads,counts))
    SAM=NULL
} 
```
