```r
computeRepIndex=function(allreads,valuevar){
    print(allreads)
    DTsam = allreads %>% spread(strand,valuevar,fill = 0)
#    DTsam=dcast.data.table(allreads,CBC+sample+datalabel~strand,value.var = valuevar,fill=0)
    DTsam = DTsam %>% mutate(readSum=(as.numeric(Negative)+as.numeric(Positive)))
    DTsam = DTsam %>% mutate(repIndex=(as.numeric(Negative)/readSum))
    DTsam = DTsam %>% mutate(DataTable=valuevar)
    return(DTsam)
}
```
